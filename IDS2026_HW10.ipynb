```python
from catboost import CatBoostRegressor
import numpy as np
import pandas as pd
from sklearn.metrics import root_mean_squared_error


def valmista_ette_catboost(df_in):
    d = df_in.copy()

    if "herd_n_milking_cows" in d.columns:
        d["log_herd_n_milking_cows"] = np.log1p(d["herd_n_milking_cows"])
        d = d.drop(columns=["herd_n_milking_cows"])
    if "herd_n_cows_total" in d.columns:
        d["log_herd_n_cows_total"] = np.log1p(d["herd_n_cows_total"])
        d = d.drop(columns=["herd_n_cows_total"])

    siht = ["target_milk_305_kg", "target_fat_305_kg", "target_protein_305_kg"]
    tunnused = [c for c in d.columns if c not in siht]

    X = d[tunnused].copy()
    y = d["target_milk_305_kg"]

    cat_cols = [c for c in ["breed", "calving_season"] if c in X.columns]
    for c in cat_cols:
        X[c] = X[c].astype(str)

    return X, y, d["calving_year"], cat_cols


X1, y1, year1, cat_cols1 = valmista_ette_catboost(parity1)
X2, y2, year2, cat_cols2 = valmista_ette_catboost(parity2)

cb1 = CatBoostRegressor(
    iterations=500, learning_rate=0.05, depth=6, random_seed=42, verbose=100
)
cb1.fit(
    X1[year1 <= 2023],
    y1[year1 <= 2023],
    cat_features=cat_cols1,
    eval_set=(X1[year1 == 2024], y1[year1 == 2024]),
    early_stopping_rounds=50,
)
preds1 = cb1.predict(X1[year1 == 2024])
rmse1 = root_mean_squared_error(y1[year1 == 2024], preds1)

print("\n" + "=" * 40 + "\n")

cb2 = CatBoostRegressor(
    iterations=500, learning_rate=0.05, depth=6, random_seed=42, verbose=100
)
cb2.fit(
    X2[year2 <= 2023],
    y2[year2 <= 2023],
    cat_features=cat_cols2,
    eval_set=(X2[year2 == 2024], y2[year2 == 2024]),
    early_stopping_rounds=50,
)
preds2 = cb2.predict(X2[year2 == 2024])
rmse2 = root_mean_squared_error(y2[year2 == 2024], preds2)

y_val_koos = pd.concat([y1[year1 == 2024], y2[year2 == 2024]])
preds_koos = np.concatenate([preds1, preds2])
koond_rmse = root_mean_squared_error(y_val_koos, preds_koos)

print("\n=== CATBOOST TULEMUSED (2024 valideerimishulgal) ===")
print(f"Esmaspoegijate (parity 1) RMSE:  {rmse1:.2f} kg")
print(f"Korduvpoegijate (parity 2+) RMSE: {rmse2:.2f} kg")
print(f"Ühine koond-RMSE (2024):          {koond_rmse:.2f} kg")
```
